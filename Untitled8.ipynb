### Load the KFC Sample Dataset and Perform EDA

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/KFC_Sample_Dataset.csv')

# Display basic information and statistics
print("Dataset Shape:", df.shape)
print("\nDataset Info:")
display(df.info())
print("\nFirst 5 Rows:")
display(df.head())
print("\nMissing Values:")
display(df.isnull().sum())
```

### Generate a Pie Chart
Depending on the dataset's columns (e.g., product categories, locations, or payment methods), we will find a categorical column and display its distribution using a pie chart.

```python
# Identify categorical columns, excluding unique identifiers like Order_ID
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
ignore_cols = ['Order_ID', 'Date', 'Time']
filtered_cat_cols = [col for col in cat_cols if col not in ignore_cols]

# Let's find a categorical column that has multiple entries per category for better visualization
if len(filtered_cat_cols) > 0:
    # Prefer Category, Payment_Mode, or Order_Type if available
    preferred = ['Category', 'Payment_Mode', 'Order_Type', 'Gender', 'City']
    target_col = next((col for col in preferred if col in filtered_cat_cols), filtered_cat_cols[0])

    print(f"Selected column for pie chart: {target_col}")

    # Prepare data for pie chart
    data_counts = df[target_col].value_counts()

    # Plotting
    plt.figure(figsize=(8, 8))
    plt.pie(data_counts, labels=data_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
    plt.title(f'Distribution of KFC Orders by {target_col}')
    plt.axis('equal')
    plt.show()
else:
    print("No suitable categorical columns found to build a pie chart.")
```

### Implement Decision Tree Classifier
We will prepare the features and target variable from the KFC dataset, encode categorical variables, and train a decision tree model.

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt

# Prepare features and target
# Let's predict 'Order_Type' based on 'Age_Group', 'Gender', 'Quantity', and 'Category'
features = ['Age_Group', 'Gender', 'Quantity', 'Category']
target = 'Order_Type'

X = df[features].copy()
y = df[target].copy()

# Encode categorical features
label_encoders = {}
for col in X.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    X[col] = le.fit_transform(X[col])
    label_encoders[col] = le

# Encode target variable
target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)

# Train Decision Tree Classifier
clf = DecisionTreeClassifier(random_state=42, max_depth=3)
clf.fit(X, y_encoded)

print("Model trained successfully.")
```

### Visualize the Decision Tree

```python
plt.figure(figsize=(12, 8))
plot_tree(clf,
          feature_names=features,
          class_names=target_encoder.classes_,
          filled=True,
          rounded=True)
plt.title("Decision Tree Structure")
plt.show()
```

### Evaluate Decision Tree Performance
We will predict the classes on our dataset and evaluate the model using Accuracy and a Classification Report (Precision, Recall, F1-Score).

```python
from sklearn.metrics import accuracy_score, classification_report

# Make predictions
y_pred = clf.predict(X)

# Calculate accuracy
accuracy = accuracy_score(y_encoded, y_pred)
print(f"Model Accuracy: {accuracy * 100:.1f}%\n")

# Print classification report
print("Classification Report:")
print(classification_report(y_encoded, y_pred, target_names=target_encoder.classes_))
```

### Export and Download the Trained Model
We will packet the decision tree model along with the feature and target encoders using `joblib` so that they can be used for future predictions, and download the file.

```python
import joblib
from google.colab import files

# Create a dictionary containing the model and all associated encoders
model_artifacts = {
    'model': clf,
    'label_encoders': label_encoders,
    'target_encoder': target_encoder,
    'features': features
}

# Save the artifacts to a file
model_filename = 'kfc_decision_tree_model.joblib'
joblib.dump(model_artifacts, model_filename)
print(f"Model and encoders saved to {model_filename}")

# Trigger browser download
files.download(model_filename)
```
